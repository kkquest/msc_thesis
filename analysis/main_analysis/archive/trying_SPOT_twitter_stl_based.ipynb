# Standard DSPOT and its modifications

In this notebook, we explore the modifications of EVT-based streaming anomaly detectors - originally proposed in the SPOT/DSPOT framework (Siffer at all, 2017) - when applied to the realTraffic time series. 

Our goals are twofold:
1. Assess the impact of a preprocessing pipeline (seasonal-trend decomposition and variance normalization) on the series’ dependence structure, so as to better satisfy the “weak dependence” assumptions required by Peaks-Over-Threshold methods.
2. Compare four variants of the DSPOT algorithm to determine which yields the most robust and accurate detection on network traffic data:
    - Vanilla DSPOT, as described in Siffer et al. (2017);
    - Preprocessed DSPOT, operating on STL residuals followed by min–max normalization of the series;
    - Binned DSPOT, which replaces the continuous GPD fit in the POT step with a histogram-based Pareto approximation to improve numerical stability;
    - Preprocessed + Binned DSPOT, combining both enhancements.

Streaming POT (SPOT) and its drift-aware extension DSPOT work by (1) calibrating an initial exceedance threshold via a high-quantile GPD fit over an initial batch, then (2) updating that threshold online by incorporating new “peaks” (excesses above the calibration threshold), while flagging any points that exceed the dynamically adjusted bound as anomalies. DSPOT further accounts for slow, deterministic changes in the baseline by subtracting a moving-window local average before applying SPOT’s logic.

Throughout this notebook, we will:
1. Preprocess the raw series via STL decomposition to remove diurnal/weekly seasonality, then normalize to the [0, 1] range.
2. Implement each of the four DSPOT variants.
3. Evaluate detection performance in terms of false-positive rate, false-negative, true-positive rate, f1 score, recall and precision, highlighting when and why certain modifications improve robustness or accuracy in practice.

This analysis allows us to form a clear picture of how preprocessing and alternative POT fitting strategies can enhance the core DSPOT framework for EVT-based anomaly detection in real-world streaming data.

## EVT‐Readiness Pipeline: Moving‐Average vs. STL-Residual Series

Before fitting any tail model or running DSPOT, we want to compare how two different detrending approaches affect our data’s dependence structure and tail behavior:

1. **Moving‐Average Series** (raw data after subtracting a depth-300 moving average)  
2. **Preprocessed Series** (STL residuals + var normalisation)

For each series, we produce the same four‐panel diagnostic figure:

| **A** | **Autocorrelation Function (ACF)** with Ljung–Box p-value annotated 

WHY? Checks remaining serial dependence.  
- **Moving‐Average** data should have reduced daily autocorrelation but may still show residual structure.  
- **Preprocessed**  should drive ACF bars near zero and yield a better p-value.
  
| **B** | **Mean Residual Life (MRL) Plot**  
>$\displaystyle \bar e(u)=\frac{1}{n_u}\sum_{x_i>u}(x_i - u)$ vs.\ threshold $u$

WHY? Guides choice of POT threshold $u$ where excesses follow GPD/Pareto behavior.  
- A linear segment indicates a range where the GPD/Pareto model is plausible.  
- Comparing curves shows whether normalisation and seasonal‐trend removal stabilise the MRL region.
  
| **C** | **QQ-Plot of GPD Exceedances** (with KS p-value)

WHY? Assesses fit of the continuous GPD to exceedances above $u$.  
- Points near the diagonal + p-value > 0.05 ⇒ no evidence against the GPD tail.  
- Differences between series reveal which detrending yields a more GPD-like tail.
  
| **D** | **QQ-Plot of Pareto Exceedances**  (histogram-Pareto approximation) 

WHY? Validates the discrete Pareto fit used by the “binned” variant.  
- Closer adherence to the 45° line suggests the histogram-Pareto is a good practical alternative to GPD for streaming. |


By applying this pipeline to both the moving-average and STL-residual series, we’ll clearly see how each detrending strategy impacts stationarity, threshold selection, and tail-fit quality, providing a rigorous foundation before any DSPOT‐based anomaly detection.

```python
# Preload timestamps & anomaly windows
timestamps  = pd.to_datetime(df2.iloc[:, 0])

# Collect metrics per depth
metrics = {}   

for ax, depth in zip(axes, depths):
    # Run DSPOT
    model = spot.dSPOT(q=proba, depth=depth)
    model.fit(init_data=init_data, data=data)
    model.initialize()
    results = model.run()
    
    # Map alarms to timestamps
    alarm_offsets  = np.array(results["alarms"])
    detected_times = timestamps[init_data + alarm_offsets]

    # Compute metrics
    window_hits = [False] * len(anomaly_windows)
    FP = 0
    for ts in detected_times:
        matched = False
        for i, (start, end) in enumerate(anomaly_windows):
            if start <= ts <= end:
                window_hits[i] = True
                matched = True
                break
        if not matched:
            FP += 1

    TP = sum(window_hits)
    FN = len(anomaly_windows) - TP
    precision = TP / (TP + FP) if (TP + FP) > 0 else 0
    recall    = TP / (TP + FN) if (TP + FN) > 0 else 0
    f1        = (2 * precision * recall / (precision + recall)
                 if (precision + recall) > 0 else 0)

    # Store metrics
    metrics[depth] = {
        "TP": TP, "FP": FP, "FN": FN,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }

# Print summary tables
for depth, m in metrics.items():
    print(f"\n=== depth = {depth} ===")
    print(f"  TP        : {m['TP']}")
    print(f"  FP        : {m['FP']}")
    print(f"  FN        : {m['FN']}")
    print(f"  Precision : {m['precision']:.3f}")
    print(f"  Recall    : {m['recall']:.3f}")
    print(f"  F1 Score  : {m['f1']:.3f}")
```

### DSPOT Extensions for Stationary Residuals

While the moving‐average detrending produces perfect event-based scores, it does not fully remove non-stationary components from our series — as shown in the EVT-suitability analysis (realTraffic_EDA_and_EVT_fit.ipynb). That residual non-stationarity violates a core EVT assumption.

To address this, we will try two extensions:
1. Standardised residuals: compute the z-score of each point—i.e. subtract a rolling mean and divide by a rolling standard deviation over a suitably long window—to stabilize both the mean and variance. We’ll feed these standardized residuals into DSPOT so that any extreme deviations stand out against a unit-variance baseline.
2. STL decomposition: apply a robust STL fit (with the appropriate seasonal period) to remove trend and seasonal components entirely, leaving only the remainder series. Because STL explicitly separates seasonality and trend, the residuals should be closer to white noise while preserving the true extremes DSPOT needs to flag.
   
Wee will evaluate these approaches side-by-side to determine which yields the best compromise between strict stationarity and extreme-value preservation.

```python
# %%  
# 1) Definition of the fixed BinnedDSPOT  
import numpy as np  
import matplotlib.pyplot as plt  

class BinnedDSPOT:  
    def __init__(  
        self,  
        q=1e-4,  
        init_data=None,  
        num_bins=50,  
        t_quantile=0.98,  
        window_size=200,  
        min_bin_count=5,  
        xmin_floor=1e-3,  
        force_min_alpha=0.5,  
        use_equipopulated=False  
    ):  
        self.q = q  
        self.num_bins = num_bins  
        self.window_size = window_size  
        self.min_bin_count = min_bin_count  
        self.xmin_floor = xmin_floor  
        self.force_min_alpha = force_min_alpha  
        self.use_equipopulated = use_equipopulated  

        self.t = None  
        self.bin_edges = None  
        self.counts = None  
        self.N_t = 0  
        self.n = 0  
        self.alpha = None  
        self.xmin = None  
        self.z_q = None  

        self.window = []  

        if init_data is not None:  
            self.fit_initial(init_data, t_quantile)  

    def fit_initial(self, data, t_quantile=0.98):  
        self.n = len(data)  
        self.t = np.quantile(data, t_quantile)  
        excesses = data[data > self.t] - self.t  
        self.N_t = len(excesses)  
        if self.N_t < 1:  
            raise ValueError("Not enough peaks above threshold for initial fit.")  

        if self.use_equipopulated:  
            sorted_ex = np.sort(excesses)  
            qs = np.linspace(0, 1, self.num_bins + 1)  
            self.bin_edges = np.unique(np.quantile(sorted_ex, qs))  
        else:  
            raw_xmin = excesses.min()  
            raw_xmax = excesses.max()  
            self.bin_edges = np.logspace(  
                np.log10(raw_xmin), np.log10(raw_xmax), self.num_bins + 1  
            )  

        self.counts, _ = np.histogram(excesses, bins=self.bin_edges)  
        self._estimate_tail()  
        self._update_threshold()  

    def _estimate_tail(self):  
        counts = self.counts + 1  
        mids = np.sqrt(self.bin_edges[:-1] * self.bin_edges[1:])  
        mask = counts >= self.min_bin_count  
        if not np.any(mask):  
            raise RuntimeError("No bins have enough counts for tail estimation.")  

        x = np.log(mids[mask])  
        y = np.log(counts[mask])  
        A = np.vstack([x, np.ones_like(x)]).T  
        slope, _ = np.linalg.lstsq(A, y, rcond=None)[0]  
        raw_alpha = -slope  

        self.alpha = max(raw_alpha, 1)  
        raw_xmin = mids[mask][0]  
        self.xmin = max(raw_xmin, self.xmin_floor)  

    def _update_threshold(self):  
        factor = (self.N_t / (self.q * self.n)) ** (1 / self.alpha)  
        self.z_q = self.t + self.xmin * factor - self.xmin  


    def update(self, x_new):  
        self.window.append(x_new)  
        if len(self.window) > self.window_size:  
            self.window.pop(0)  
        M = np.mean(self.window)  

        x_resid = x_new - M  
        self.n += 1  

        #if x_resid > self.z_q:  # compare to the residuals 
         #   return True  
        if x_new > (M + self.z_q): # compare to the raw value
            return True


        if x_resid > self.t:  
            y = x_resid - self.t  
            idx = np.searchsorted(self.bin_edges, y, side='right') - 1  
            if 0 <= idx < len(self.counts):  
                self.counts[idx] += 1  
                self.N_t += 1  
                self._estimate_tail()  
                self._update_threshold()  

        return False  

# %%  
# 2) Load your df2 and extract values/timestamps  
#    (adjust column names if needed)  
df2 = pd.read_csv("../data/raw_data/Twitter_volume_AMZN.csv", parse_dates=[0])
df2.columns = ["timestamp", "value"]

timestamps = df2['timestamp'].values  
values     = df2['value'].values  

# %%  
# 3) Instantiate detector with a larger init batch  
init_size = 5000  
detector = BinnedDSPOT(  
    q=1e-4,  
    init_data=values[:init_size],  
    num_bins=20,  
    t_quantile=0.98,  
    window_size=200,  
    min_bin_count=5,  
    xmin_floor=1e-3,  
    force_min_alpha=0.5,  
    use_equipopulated=True  
)  

# ← Seed the sliding window and sample count BEFORE streaming  
detector.window = list(values[init_size - detector.window_size : init_size])  
detector.n = init_size  


# 4) Stream and collect anomalies + threshold history  
anomaly_idxs  = []  
anomaly_vals  = []  
z_history     = []      # residual-space thresholds  
M_history     = []      # local means  

for i, x in enumerate(values[init_size:], start=init_size):  
    # compute and store current window mean  
    M = np.mean(detector.window) if detector.window else 0  
    M_history.append(M)
    
    # original update only returns True/False, but we use the new logic:
    detector.window.append(x)
    if len(detector.window) > detector.window_size:
        detector.window.pop(0)
    x_resid = x - M
    detector.n += 1

    # flag by raw value
    if x > M + detector.z_q:
        anomaly_idxs.append(i)
        anomaly_vals.append(x)
    elif x_resid > detector.t:
        # peak update as before
        y = x_resid - detector.t
        idx = np.searchsorted(detector.bin_edges, y, side='right') - 1
        if 0 <= idx < len(detector.counts):
            detector.counts[idx] += 1
            detector.N_t += 1
            detector._estimate_tail()
            detector._update_threshold()

    z_history.append(detector.z_q)


print(f"Detected {len(anomaly_vals)} anomalies out of {len(values)-init_size} samples")  
# %%  
# 5) Build results DataFrame and show top anomalies  
import pandas as pd  
results = pd.DataFrame({  
    'index':     anomaly_idxs,  
    'timestamp': timestamps[anomaly_idxs],  
    'value':     anomaly_vals  
})  
display(results.head())  


# 6) Plot stream, anomalies, and **raw** threshold  
plt.figure(figsize=(12,4))  
plt.plot(timestamps[init_size:], values[init_size:], label='Value')  
plt.scatter(timestamps[anomaly_idxs], anomaly_vals, s=10, color='red', label='Anomalies')  

# compute raw thresholds = M_history + z_history
raw_thresh = np.array(M_history) + np.array(z_history)
plt.plot(timestamps[init_size:], raw_thresh, label='raw-space threshold')  

plt.xlabel('Time')  
plt.ylabel('Value')  
plt.title('Binned-DSPOT on df2 (raw threshold)')  
plt.legend()  
plt.tight_layout()  
plt.show()

```

```python
from itertools import product
import pandas as pd
import numpy as np

# 0) Prepare datetime windows once
anomaly_windows = [
    (pd.to_datetime(s), pd.to_datetime(e))
    for s, e in anomaly_windows
]

# 1) Hyper-parameter grid
init_sizes    = [1000, 2000, 5000]
num_bins_list = [20, 30, 50, 100]
min_counts    = [1, 2, 5]
alphas        = [0.5, 1.0]    # force_min_alpha

results = []
for init_size, nb, mc, fa in product(init_sizes, num_bins_list, min_counts, alphas):
    if init_size >= len(values):
        continue

    # 2) Initialize detector and seed its window + sample count
    try:
        det = BinnedDSPOT(
            q=1e-4,
            init_data=values[:init_size],
            num_bins=nb,
            t_quantile=0.98,
            window_size=200,
            min_bin_count=mc,
            xmin_floor=1e-3,
            force_min_alpha=fa,
            use_equipopulated=True
        )
        det.window = list(values[init_size - det.window_size : init_size])
        det.n = init_size
    except Exception:
        continue

    # 3) Stream with manual ordering
    anomaly_idxs = []
    anomaly_vals = []
    for i, x in enumerate(values[init_size:], start=init_size):
        # 3.1) Compute M on previous window
        M = np.mean(det.window)
        # 3.2) Flag in raw space
        if x > M + det.z_q:
            anomaly_idxs.append(i)
            anomaly_vals.append(x)
        else:
            # 3.3) Peak update on residual
            x_resid = x - M
            if x_resid > det.t:
                y = x_resid - det.t
                idx = np.searchsorted(det.bin_edges, y, side="right") - 1
                if 0 <= idx < len(det.counts):
                    det.counts[idx] += 1
                    det.N_t += 1
                    det._estimate_tail()
                    det._update_threshold()
        # 3.4) Update window and sample count for next iter
        det.window.append(x)
        if len(det.window) > det.window_size:
            det.window.pop(0)
        det.n += 1

    # 4) Event-based scoring
    TP = FP = 0
    hits = [False] * len(anomaly_windows)
    for idx in anomaly_idxs:
        ts = timestamps[idx]
        matched = False
        for j, (start, end) in enumerate(anomaly_windows):
            if start <= ts <= end:
                hits[j] = True
                matched = True
                break
        if not matched:
            FP += 1
    TP = sum(hits)
    FN = len(anomaly_windows) - TP
    prec = TP / (TP + FP) if TP + FP else 0.0
    rec  = TP / (TP + FN) if TP + FN else 0.0
    f1   = (2 * prec * rec / (prec + rec)) if (prec + rec) else 0.0

    results.append({
        'init_size': init_size,
        'num_bins': nb,
        'min_count': mc,
        'min_alpha': fa,
        'TP': TP, 'FP': FP, 'FN': FN,
        'precision': prec, 'recall': rec, 'f1': f1
    })

# 5) Summarize top configurations by F1
df_scores = pd.DataFrame(results).sort_values('f1', ascending=False)
display(df_scores.head())
```

```python
# binned dspot, works but plots the residul threshold 
# %%  
# 1) Definition of the fixed BinnedDSPOT  
import numpy as np  
import matplotlib.pyplot as plt  

class BinnedDSPOT:  
    def __init__(  
        self,  
        q=1e-4,  
        init_data=None,  
        num_bins=50,  
        t_quantile=0.98,  
        window_size=200,  
        min_bin_count=5,  
        xmin_floor=1e-3,  
        force_min_alpha=0.5,  
        use_equipopulated=False  
    ):  
        self.q = q  
        self.num_bins = num_bins  
        self.window_size = window_size  
        self.min_bin_count = min_bin_count  
        self.xmin_floor = xmin_floor  
        self.force_min_alpha = force_min_alpha  
        self.use_equipopulated = use_equipopulated  

        self.t = None  
        self.bin_edges = None  
        self.counts = None  
        self.N_t = 0  
        self.n = 0  
        self.alpha = None  
        self.xmin = None  
        self.z_q = None  

        self.window = []  

        if init_data is not None:  
            self.fit_initial(init_data, t_quantile)  

    def fit_initial(self, data, t_quantile=0.98):  
        self.n = len(data)  
        self.t = np.quantile(data, t_quantile)  
        excesses = data[data > self.t] - self.t  
        self.N_t = len(excesses)  
        if self.N_t < 1:  
            raise ValueError("Not enough peaks above threshold for initial fit.")  

        if self.use_equipopulated:  
            sorted_ex = np.sort(excesses)  
            qs = np.linspace(0, 1, self.num_bins + 1)  
            self.bin_edges = np.unique(np.quantile(sorted_ex, qs))  
        else:  
            raw_xmin = excesses.min()  
            raw_xmax = excesses.max()  
            self.bin_edges = np.logspace(  
                np.log10(raw_xmin), np.log10(raw_xmax), self.num_bins + 1  
            )  

        self.counts, _ = np.histogram(excesses, bins=self.bin_edges)  
        self._estimate_tail()  
        self._update_threshold()  

    def _estimate_tail(self):  
        counts = self.counts + 1  
        mids = np.sqrt(self.bin_edges[:-1] * self.bin_edges[1:])  
        mask = counts >= self.min_bin_count  
        if not np.any(mask):  
            raise RuntimeError("No bins have enough counts for tail estimation.")  

        x = np.log(mids[mask])  
        y = np.log(counts[mask])  
        A = np.vstack([x, np.ones_like(x)]).T  
        slope, _ = np.linalg.lstsq(A, y, rcond=None)[0]  
        raw_alpha = -slope  

        self.alpha = max(raw_alpha, 1)  
        raw_xmin = mids[mask][0]  
        self.xmin = max(raw_xmin, self.xmin_floor)  

    def _update_threshold(self):  
        factor = (self.N_t / (self.q * self.n)) ** (1 / self.alpha)  
        self.z_q = self.t + self.xmin * factor - self.xmin  

    def update(self, x_new):  
        self.window.append(x_new)  
        if len(self.window) > self.window_size:  
            self.window.pop(0)  
        M = np.mean(self.window)  

        x_resid = x_new - M  
        self.n += 1  

        #if x_resid > self.z_q:  # compare to the residuals 
         #   return True  
        if x_new > (M + self.z_q): # compare to the raw value
            return True

        if x_resid > self.t:  
            y = x_resid - self.t  
            idx = np.searchsorted(self.bin_edges, y, side='right') - 1  
            if 0 <= idx < len(self.counts):  
                self.counts[idx] += 1  
                self.N_t += 1  
                self._estimate_tail()  
                self._update_threshold()  

        return False  

# %%  
# 2) Load your df2 and extract values/timestamps  
#    (adjust column names if needed)  
timestamps = df2['timestamp'].values  
values     = df2['value'].values  

# %%  
# 3) Instantiate detector with a larger init batch  
init_size = 5000  
detector = BinnedDSPOT(  
    q=1e-4,  
    init_data=values[:init_size],  
    num_bins=20,  
    t_quantile=0.98,  
    window_size=200,  
    min_bin_count=5,  
    xmin_floor=1e-3,  
    force_min_alpha=0.5,  
    use_equipopulated=True  
)  
# %%  
# 4) Stream and collect anomalies + threshold history  
anomaly_idxs  = []  
anomaly_vals  = []  
z_history     = []  

for i, x in enumerate(values[init_size:], start=init_size):  
    z_history.append(detector.z_q)  
    if detector.update(x):  
        anomaly_idxs.append(i)  
        anomaly_vals.append(x)  

print(f"Detected {len(anomaly_vals)} anomalies out of {len(values)-init_size} samples")  


# %%  
# 5) Build results DataFrame and show top anomalies  
import pandas as pd  
results = pd.DataFrame({  
    'index':     anomaly_idxs,  
    'timestamp': timestamps[anomaly_idxs],  
    'value':     anomaly_vals  
})  
display(results.head())  

# %%  
# 6) Plot stream, anomalies, and threshold  
plt.figure(figsize=(12,4))  
plt.plot(timestamps[init_size:], values[init_size:], label='Value')  
plt.scatter(timestamps[anomaly_idxs], anomaly_vals, s=10, color='red', label='Anomalies')  
plt.plot(timestamps[init_size:], z_history, label='z_q Threshold')  
plt.xlabel('Time')  
plt.ylabel('Value')  
plt.title('Binned-DSPOT on df2')  
plt.legend()  
plt.tight_layout()  
plt.show()  
```
