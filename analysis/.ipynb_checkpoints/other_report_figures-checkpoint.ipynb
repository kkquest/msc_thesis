```python
# === EVT anomaly illustrations: tail vs bulk ===
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Output paths
out_dir = Path("../output/others") 
out_dir.mkdir(parents=True, exist_ok=True) 

tail_path = out_dir / "evt_tail_anomaly.png"
bulk_path = out_dir / "evt_bulk_anomaly.png"

# Tail anomaly 
x = np.linspace(-4, 8, 800)
normal_pdf = (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * x**2)

# pick a far-right tail location to mark as an "anomaly"
tail_x = 5.5
tail_y = (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * tail_x**2)

# Bulk anomaly 
x2 = np.linspace(-6, 6, 900)
# bimodal mixture: N(-2,1) and N(2,1), equal weights
bimodal_pdf = 0.5 * (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * (x2 + 2)**2) \
            + 0.5 * (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * (x2 - 2)**2)

# "Bulk anomaly" at the valley between modes
bulk_x = 0.0
bulk_y = 0.5 * (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * (bulk_x + 2)**2) \
       + 0.5 * (1 / np.sqrt(2 * np.pi)) * np.exp(-0.5 * (bulk_x - 2)**2)

# Tail plot
fig1, ax1 = plt.subplots(figsize=(6, 3.2))
ax1.plot(x, normal_pdf, linewidth=2)
ax1.axvline(tail_x, linestyle='--', linewidth=1.5)
ax1.scatter([tail_x], [tail_y], s=60, marker='*')
ax1.text(tail_x + 0.15, tail_y, "Tail anomaly", fontsize=10, va="bottom")
ax1.set_title("(a) Tail anomaly (EVT applies)")
ax1.set_xlabel("x")
ax1.set_ylabel("Density")
ax1.grid(alpha=0.3)
fig1.tight_layout()
fig1.savefig(tail_path, dpi=300)
plt.close(fig1)

# Bulk plot
fig2, ax2 = plt.subplots(figsize=(6, 3.2))
ax2.plot(x2, bimodal_pdf, linewidth=2)
ax2.axvline(bulk_x, linestyle='--', linewidth=1.5)
ax2.scatter([bulk_x], [bulk_y], s=60, marker='*')
ax2.text(bulk_x + 0.15, bulk_y, "Bulk anomaly", fontsize=10, va="bottom")
ax2.set_title("(b) Bulk anomaly in a bimodal distribution (EVT does not apply)")
ax2.set_xlabel("x")
ax2.set_ylabel("Density")
ax2.grid(alpha=0.3)
fig2.tight_layout()
fig2.savefig(bulk_path, dpi=300)
plt.close(fig2)

print("Saved:", tail_path, bulk_path, sep="\n- ")
```
